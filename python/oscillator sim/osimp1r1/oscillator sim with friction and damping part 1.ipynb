```python
#this will be a collaborative simulation of a harmonic oscillator, that includes friction and damping, with Saatvik K

#all the stuff down here has been deleted go to github to see what was here before

#Round 1: No damping or friction
#Before we start varying the parameters, let's start with m=3 kg, k=108 N/m, and a time interval of [0,10] (seconds).
```

```python
%matplotlib inline
%matplotlib notebook

import numpy as np
import matplotlib.pyplot as plt

#values
N=1000
h=.01
num_cycles=30/np.pi
m=3
k=108

t=np.zeros(N)
t[0]=0

x=np.zeros(N)
x[0]=1

v=np.zeros(N)
v[0]=1

eq_state = np.array([x[0],v[0]])

f=np.zeros(N) #analytic
f[0]=1

f_prime=np.zeros(N) #analytic
f_prime[0]=1

K=np.zeros(N) #energies
U=np.zeros(N)
E=np.zeros(N)
e_loss=[]

max_amps=[]
min_amps=[]
```

```python
%matplotlib inline
%matplotlib notebook

#functions
def diff(t,state):
    [x,v]=state
    dx = v
    dv = (-k/m)*x
    return np.array([dx,dv])

#this step actually gives the updated state from rk4
def rk4(t,state):
    k1 = diff(t,
              state)
    k2 = diff(t+h/2,
              state+(h*k1/2))
    k3 = diff(t+h/2,
              state+(h*k2/2))
    k4 = diff(t+h,
              state+(h*k3))
    return state+(h/6)*(k1+2*k2+2*k3+k4)

#adding terms
for n in range(N-1):
    eq_state=(rk4(t[n],eq_state))

    t[n+1]=t[n]+h
    x[n+1]=eq_state[0]
    v[n+1]=eq_state[1]

    f[n+1]= np.cos(6*t[n+1])+(1/6)*np.sin(6*t[n+1])
    f_prime[n+1]= -6*np.sin(6*t[n+1])+np.cos(6*t[n+1])

for n in range(N):
    K[n]=(1/2)*m*(v[n])**2
    U[n]=(1/2)*k*(x[n])**2
    E[n]=K[n]+U[n]

    #the period is 2pi * sqrt(m/k) = 2pi * sqrt(1/36) = pi/3, so the number of cycles in 10 seconds is N = 10/pi/3 = 30/pi

#one cycle every second
#time list is 1000 items
#time partitions: 0->99 100->199 200->299 300->399 400->499 500->599 600->699 700->799 800->899 900->999

for n in range(int(np.round(num_cycles,decimals=0))):
    e_loss.append(-E[99+(n*100)]+E[0+(n*100)])

for n in range(int(np.round(num_cycles,decimals=0))):
    max_amps.append(max(x[(0+(n*100)):(100+(n*100))]))
```

```python
%matplotlib inline
%matplotlib notebook

#plotting RK4 solutions
plt.figure()
plt.plot(t,x,linewidth=4)
plt.title("Position vs. Time (RK4)", fontsize=24)
plt.xlabel("Time",fontsize=14)
plt.ylabel("Position",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_pvtr.png")
plt.show()

plt.figure()
plt.plot(t,v,linewidth=4)
plt.title("Velocity vs. Time (RK4)", fontsize=24)
plt.xlabel("Time",fontsize=14)
plt.ylabel("Velocity",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_vvtr.png")
plt.show()

plt.figure()
plt.plot(t,K,linewidth=4)
plt.title("Kinetic Energy vs. Time (RK4)", fontsize=24)
plt.xlabel("Time",fontsize=14)
plt.ylabel("Kinetic Energy",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_kvtr.png")
plt.show()

plt.figure()
plt.plot(t,U,linewidth=4)
plt.title("Elastic Potential Energy vs. Time (RK4)", fontsize=24)
plt.xlabel("Time",fontsize=14)
plt.ylabel("Elastic Potential Energy",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_uvtr.png")
plt.show()

plt.figure()
plt.plot(t,E,linewidth=4)
plt.title("Total Mechanical Energy vs. Time (RK4)", fontsize=24)
plt.xlabel("Time",fontsize=14)
plt.ylabel("Total Mechanical Energy",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_evtr.png")
plt.show()

plt.figure()
for n in range(len(e_loss)):
    plt.scatter(n+1,e_loss[n],linewidth=4,color="red")
plt.title("Energy Loss vs. Time", fontsize=24)
plt.xlabel("Cycle #",fontsize=14)
plt.ylabel("Energy Loss",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_elvcr.png")
plt.show()

plt.figure()
for n in range(len(e_loss)):
    plt.scatter(n+1,max_amps[n],linewidth=4,color="red")
plt.title("Max Amps/Decay Envelope vs. Time", fontsize=24)
plt.xlabel("Cycle #",fontsize=14)
plt.ylabel("Max Amp in Cycle",fontsize=14)
plt.xlim()
plt.ylim()
plt.axhline(y=0, color='k')
plt.axvline(x=0, color='k')
plt.savefig("osim_p1r1_mavcr.png")
plt.show()

#the energy loss is very miniscule (constantly 3.5*10^-6), which is good and shows RK4 has a very small error on the order of 10^-6
```
